```python
#import libraries for calculations and data manipulation
import numpy as np
import pandas as pd
```

```python
#input data
data = pd.Series([1000, 500, 800, 700, 700, 800, 700, 900, 750, 850])

#sort data
data = data.sort_values()
print(f"sorted_data: {data.to_list()}")
print("\n", data.describe())
```

```python
#get the Q1, Q3 and IQR
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1
print(f"Q1: {Q1}, Q3: {Q3}")
print(f"IQR: {IQR}")
```

```python
#get the lower and upper bound values to check if there are any outliers
lower_whisker = Q1 - (1.5 * IQR)
upper_whisker = Q3 + (1.5 * IQR)
print(f"Lower_bound: {lower_whisker}, Upper_bound: {upper_whisker}")
```

```python
#import library for visualisation
import seaborn as sns
```

```python
#generate boxplot
sns.boxplot(data)
```

```python
#calculate z-score to 
z_score = (data - data.mean())/data.std(ddof=0)
print(f"mean: {data.mean()}")
print(f"\nstandard deviation: {data.std(ddof=0)}")
print(f"\nz_score: \n{z_score.sort_values()}")
```

**Interpretation**
- The `boxplot` and `z scores` indicate there are no outliers in the dataset. The z-score values range from approximately -2.1 to 1.8 which shows that the values are within 3 standard deviations below and above the mean.
